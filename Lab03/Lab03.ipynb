# T2(h) explanation

The Price column contains text values instead of numeric values.
Some prices include the text "Rs" and commas, so Pandas cannot
directly compare them with the number 1,000,000. We need to
clean and convert the Price column into a numeric data type first.

## Task 3 · Clean the data: the six types of problem
Fix the problems **in this order** (text first, then duplicates):
- **(a) Missing values:** count them per column. Drop ads with no `Brand` or no `Model`; fill missing `Registered City`, `Fuel` and `Transaction Type` with `"Unknown"`; leave `Year` and `KMs Driven` as NaN. Print how many rows were removed.
- **(b) Formatting:** remove extra spaces and fix the case of city names (`.str.strip().str.title()`); remove `"Rs"` and commas from `Price`.
- **(c) Data types:** convert `Price` to `int64` and `Year` to `Int64`.
- **(d) Inconsistent categories:** `"Sukkar"` → `"Sukkur"` and `"Corrolla Altis"` → `"Corolla Altis"`. Print the number of different cities before and after.
- **(e) Duplicates:** count exact duplicate ads, drop them, print the new row count.
- **(f) Invalid values:** set `KMs Driven` to NaN when it is below 100, above 1,000,000, or the placeholder 12345 / 123456. Drop ads with the fake prices 12,345,678 or 87,654,321, and Suzuki/Daihatsu ads above 5,000,000.
- **(g) Self-check:** your final shape must be **(20484, 9)**.

```python


# T3a: Missing Values

print("T3a - Missing Values Before Cleaning:")
print(df.isnull().sum())

rows_before = len(df)

# Drop rows with missing Brand or Model
df = df.dropna(subset=["Brand", "Model"]).copy()

rows_removed = rows_before - len(df)

# Fill missing categorical values
df["Registered City"] = df["Registered City"].fillna("Unknown")
df["Fuel"] = df["Fuel"].fillna("Unknown")
df["Transaction Type"] = df["Transaction Type"].fillna("Unknown")

print("Rows Removed:", rows_removed)


# T3b: Formatting

# Remove extra spaces and fix city name case
df["Registered City"] = (
    df["Registered City"].str.strip().str.title()
)

# Remove Rs and commas from Price
df["Price"] = (
    df["Price"]
    .astype(str)
    .str.replace("Rs", "", regex=False)
    .str.replace(",", "", regex=False)
    .str.strip()
)

print("\nT3b - Formatting Completed")


# T3c: Data Types

df["Price"] = pd.to_numeric(df["Price"]).astype("int64")

df["Year"] = pd.to_numeric(
    df["Year"], errors="coerce"
).astype("Int64")

print("\nT3c - Updated Data Types:")
print(df.dtypes)


# T3d: Inconsistent Categories

cities_before = df["Registered City"].nunique()

df["Registered City"] = df["Registered City"].replace({
    "Sukkar": "Sukkur"
})

df["Model"] = df["Model"].replace({
    "Corrolla Altis": "Corolla Altis"
})

cities_after = df["Registered City"].nunique()

print("\nT3d - Different Cities Before:", cities_before)
print("T3d - Different Cities After:", cities_after)


# T3e: Duplicates

duplicate_count = df.duplicated().sum()

print("\nT3e - Exact Duplicate Ads:", duplicate_count)

df = df.drop_duplicates().copy()

print("Rows After Removing Duplicates:", len(df))


# T3f: Invalid Values

# Convert KMs Driven to numeric
df["KMs Driven"] = pd.to_numeric(
    df["KMs Driven"], errors="coerce"
)

# Find invalid kilometre values
invalid_km = (
    (df["KMs Driven"] < 100) |
    (df["KMs Driven"] > 1000000) |
    (df["KMs Driven"].isin([12345, 123456]))
)

# Replace invalid values with NaN
df.loc[invalid_km, "KMs Driven"] = np.nan

# Remove fake prices
fake_prices = df["Price"].isin([12345678, 87654321])

# Remove unusually high Suzuki and Daihatsu prices
invalid_brand_prices = (
    df["Brand"].isin(["Suzuki", "Daihatsu"]) &
    (df["Price"] > 5000000)
)

df = df.loc[
    ~(fake_prices | invalid_brand_prices)
].copy()


# T3g: Self Check

print("\nT3g - Final Dataset Shape:")
print(df.shape)

print("\nFinal Missing Values:")
print(df.isnull().sum())

print("\nCleaning Completed!")
```

## Task 4 · Analyse
- **(a)** New column `Price_lakh` = price in lakh rupees (÷ 100,000).
- **(b)** New column `Age` = 2020 − `Year`.
- **(c)** The 5 most common brands.
- **(d)** The 5 cities with the most ads.
- **(e)** Median `Price_lakh` for each fuel type, highest first.
- **(f)** Median `Price_lakh` of a **Corolla GLI** for each year 2010–2018. How much value does it lose per year?
- **(g)** The 5 most expensive ads. Does one of them look wrong? Why?
- **(h)** Number of ads and median price for Cash vs Installment/Leasing. Interpret in 1–2 lines.

```python


# T4a: Create Price_lakh column

df["Price_lakh"] = df["Price"] / 100000

print("T4a - Price in Lakh:")
display(df[["Brand", "Price", "Price_lakh"]].head())


# T4b: Create Age column

df["Age"] = 2020 - df["Year"]

print("\nT4b - Car Age:")
display(df[["Brand", "Year", "Age"]].head())


# T4c: Five Most Common Brands

print("\nT4c - Top 5 Brands:")
display(df["Brand"].value_counts().head(5))


# T4d: Five Cities with Most Ads

print("\nT4d - Top 5 Cities:")
display(df["Registered City"].value_counts().head(5))


# T4e: Median Price for Each Fuel Type

print("\nT4e - Median Price by Fuel Type:")

fuel_median = (
    df.groupby("Fuel")["Price_lakh"]
    .median()
    .sort_values(ascending=False)
)

display(fuel_median)


# T4f: Corolla GLI Median Price from 2010 to 2018

print("\nT4f - Corolla GLI Median Price:")

corolla = df[
    (df["Model"] == "Corolla GLI") &
    (df["Year"].between(2010, 2018))
]

corolla_median = (
    corolla.groupby("Year")["Price_lakh"]
    .median()
    .reindex(range(2010, 2019))
)

display(corolla_median)

# Calculate average yearly price difference
if pd.notna(corolla_median.loc[2010]) and pd.notna(corolla_median.loc[2018]):
    yearly_loss = (
        corolla_median.loc[2010] -
        corolla_median.loc[2018]
    ) / 8

    print("Average Price Difference Per Year:", yearly_loss, "Lakh")
else:
    print("Yearly difference cannot be calculated because an endpoint is missing.")


# T4g: Five Most Expensive Ads

print("\nT4g - Five Most Expensive Cars:")

display(
    df.nlargest(5, "Price")[
        ["Brand", "Model", "Year", "Price"]
    ]
)


# T4h: Cash vs Installment/Leasing

print("\nT4h - Transaction Type Analysis:")

transaction_data = df[
    df["Transaction Type"].isin(["Cash", "Installment/Leasing"])
]

transaction_summary = (
    transaction_data
    .groupby("Transaction Type")["Price"]
    .agg(["count", "median"])
)

display(transaction_summary)
```
