## Task 7 · Climate of five cities (real data, PMD 1991–2020 normals)
Using `tmax` (mean daily maximum °C) and `rain` (mm), rows = `cities`, columns = `months`:
- **(a)** Shape of `tmax`.
- **(b)** Annual mean maximum temperature of each city.
- **(c)** Hottest month of each city (use `argmax` + `months`).
- **(d)** 5-city average for each month.
- **(e)** Number of months above 35 °C in each city.
- **(f)** Lahore's temperatures in °F (`F = C × 9/5 + 32`).
- **(g)** Monsoon (Jul–Sep) rain of each city by slicing columns.
- **(h)** The wettest city in the monsoon, and what % of each city's yearly rain falls in the monsoon.

**Interpret:** in 1–2 lines, which result surprised you, and why?

```python
print("T7a: shape of tmax =", tmax.shape)

annual_mean = tmax.mean(axis=1)
for c_, m_ in zip(cities, annual_mean):
    print(f"T7b: {c_:<10s} annual mean max = {m_:5.2f} °C")

hottest = months[tmax.argmax(axis=1)]
print("T7c: hottest month of each city =", dict(zip(cities.tolist(), hottest.tolist())))

print("T7d: 5-city monthly average (°C) =", np.round(tmax.mean(axis=0), 2))

print("T7e: months above 35 °C per city =", dict(zip(cities.tolist(), (tmax > 35).sum(axis=1).tolist())))

lahore_F = tmax[0] * 9/5 + 32
print("T7f: Lahore temperatures (°F) =", np.round(lahore_F, 1))

monsoon = rain[:, 6:9]                       # Jul, Aug, Sep
monsoon_total = monsoon.sum(axis=1)
print("T7g: monsoon rain per city (mm) =", dict(zip(cities.tolist(), np.round(monsoon_total, 1).tolist())))

print("T7h: wettest monsoon city =", cities[monsoon_total.argmax()])
pct = monsoon_total / rain.sum(axis=1) * 100
print("T7h: monsoon share of yearly rain (%) =", dict(zip(cities.tolist(), np.round(pct, 1).tolist())))
```

Surprise: Islamabad gets about 763 mm of rain in Jul-Sep, far more than Lahore (454 mm). Karachi is the hottest city on average (32.5 °C) but its hottest month is only May. Quetta gets almost no monsoon rain (just 6 % of its yearly total).

## Task 9 · Grace marks
Using `marks`:
- **(a)** Add 5 grace marks but cap at 100 (`np.clip`).
- **(b)** Pass % (≥ 50) after grace.
- **(c)** Index of students who passed **only because of** the grace marks.
- **(d)** z-score of each student after grace.

```python
after = np.clip(marks + 5, 0, 100)
print("T9a: marks after grace =", after)

print("T9b: pass % after grace =", np.mean(after >= 50) * 100, "%")

only_grace = np.where((marks < 50) & (after >= 50))[0]
print("T9c: indices passed only due to grace =", only_grace, "| marks before =", marks[only_grace])

z = (after - after.mean()) / after.std()
print("T9d: z-scores =", np.round(z, 2))
```
